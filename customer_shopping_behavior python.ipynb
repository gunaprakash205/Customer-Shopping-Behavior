```
# importing  libs

import pandas as pd
import numpy as np

#loading data to pandas.

data=pd.read_csv("customer_shopping_behavior.csv")

# viewing data to better understanding.

#print(data.head())
#print(data.info())
#print(data.describe(include="all"))

# checking for null values .

#print(data.isnull().sum())

'''we got 37 null values in review column ,
know i replace the null values with mean or median , 
    but i replace the null values with median because mean is effected by outliners.'''


# removing null values.

'''i can fill all the nulls with median of total reviews but as to become good in data handling 
i fill null with median of there own category to get more accurate results .'''

data["Review Rating"]=data.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.median()))

#checking null values filled or not.

#print(data.isnull().sum())   # we see that no values are null in data.

'''when i output i notice that names of columns are
 not that ordered so we change that into snake format to  easy access of columns '''

# organize the data .

data.columns=data.columns.str.lower()

data.columns=data.columns.str.replace(' ',"_")

data.columns=data.columns.str.replace('purchase_amount_(usd)','purchase_amount')

#print(data.columns)
labels=["young adults","adults","older","senior"]
data["age_group"]=pd.qcut(data["age"],q=4,labels=labels)

#print(data.head(10))
freq_pur={
    "Fortnightly":14,
        "Weekly":7,
        "Annually":365,
        "Quarterly":90,
        "Monthly":30,
    "Every 3 Months":90,
    "Bi-Weekly":14
}
data["frequency_in_no"]=data["frequency_of_purchases"].map(freq_pur)
#print(data.head())

'''know i see that discount column and prompt code is both same 
lets check and see if they both are same delete one column '''

#print((data["discount_applied"]==data['promo_code_used']).all())
# here we get true so they both are same delete one column
data=data.drop('promo_code_used',axis=1)
print(data.columns)


'''import pysql alchemy'''

from sqlalchemy import create_engine

username="root"
password="gunaprakash90"
port = 3306
host="localhost"
database="mydatabase"
engine=create_engine(f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}")
table_name="mytable"
data.to_sql(table_name,engine,if_exists="replace",index=False)
#data.read_sql("select * from mytable limit 5",engine)


```
